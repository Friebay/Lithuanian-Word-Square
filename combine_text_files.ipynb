```python
from pathlib import Path

data_folder = Path("data")
all_words = set()
for file_path in data_folder.rglob("lithuanian_words.txt"):
    with open(file_path, 'r', encoding='utf-8') as f:
        print(f"Reading {file_path}")
        words = f.read().split()
        all_words.update(words)
```
